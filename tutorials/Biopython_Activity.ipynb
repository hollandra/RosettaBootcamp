###Perform initial analysis of sequence

Print the following information about my_seq:



1.   The length of my_seq
2.   The % of GC content
3.   The number of times "TTT" occurs in the sequence





```python
#Modify the code here:

#Remember to convert your sequence to the correct variable type. Can check with print(type(bio_seq)) to make sure its converted.
bio_seq = Seq(my_seq_str)

#Define seq_len, GC_percent, and TTT_count here
seq_len = len(bio_seq)
GC_percent = (bio_seq.count("G") + bio_seq.count("C")) / seq_len
TTT_count = bio_seq.count("TTT")

#Uncomment to check values
print(seq_len)
print("%.2f" % GC_percent)
print(TTT_count)
```

###Modify sequence and translate

The PhD student who gave you this sequence needs the corresponding amino acid sequence. Unforunately, they also manually copied the sequence and forgot to write out the last six nucleotides. So it will need to be adjusted before it is translated:



1.   Add the six missing nucleotides of "missing_nucs" to the end of the sequence
2.   Transcribe the sequence to RNA
3.   Translate the RNA sequence to corresponding amino acid product

```python
#Complete the code here

missing_nucs = Seq("GACTGA")

#Add the missing_nucs to the end of bio_seq
new_bio_seq = bio_seq + missing_nucs


#Uncomment to check that new_bio_seq has missing nucs at the end
print("Complete Sequence:", new_bio_seq)

rna_seq = new_bio_seq.transcribe()

#Uncomment to check that new_bio_seq is an mRNA sequence
print("", rna_seq)

prot_seq = rna_seq.translate()
print("", prot_seq)
```

###Basic Protein Analysis

Look at the header information printed above. This appears to be a DNA binding protein, a class of proteins with a large range of sizes and complexity. While one can just look at a 3d visualization of the structure to gain basic information, you know that you will be processing a large number of files. So you want to complete a basic analysis using biopython to lay the groundwork to automate the process in the future



1.   How many chains are present in this pdb file? How many residues are in these chain?

2.   Print out the amino acid identities of all residues in chain A. What is the identity of the third amino acid?

3.   Which residues of chain A are further apart form one another? Residues 4 and 19 or resides 15 and 50





```python
#Execute this code block to help with solutions

three_to_one = {
    "ALA": "A",
    "ARG": "R",
    "ASN": "N",
    "ASP": "D",
    "CYS": "C",
    "GLN": "Q",
    "GLU": "E",
    "GLY": "G",
    "HIS": "H",
    "ILE": "I",
    "LEU": "L",
    "LYS": "K",
    "MET": "M",
    "PHE": "F",
    "PRO": "P",
    "SER": "S",
    "THR": "T",
    "TRP": "W",
    "TYR": "Y",
    "VAL": "V",
}
```

```python
#Complete the code below

#Assign the structure object to my_struct
parser = PDBParser()
my_struct = parser.get_structure("enh_protein","1ENH.pdb")

#To make things easier, you can assume that there is only one model in this structure
model = my_struct[0]

#Count the number of chains in model. This can be accomplished with a for loop or directly accessing the length of a certain variable
chain_num = len(model)
print(chain_num)


#Complete the code below to print out the amino acid identities of chain A:
chain_A = model['A']
sequence = ""

for residue in chain_A:

  #Just fill out this line
  current_residue = residue.get_resname()

  if current_residue in three_to_one:
    sequence += three_to_one[current_residue]

print(sequence)

#Complete the code below to calculate the distances of central carbon atoms of residues 4,19 and 15,50

#Correctly define residues
residue_4 = my_struct[0]["A"][3]
residue_19 = my_struct[0]["A"][18]
residue_15 = my_struct[0]["A"][14]
residue_50 = my_struct[0]["A"][49]

#Correctly define central carbon atoms
atom_4 = my_struct[0]["A"][3]["CA"]
atom_19 = my_struct[0]["A"][18]["CA"]
atom_15 = my_struct[0]["A"][14]["CA"]
atom_50 = my_struct[0]["A"][49]["CA"]

#Calculate distances of central carbon atom pairs

distance_pair_1 = atom_4 - atom_19
distance_pair_2 = atom_15 - atom_50

print(distance_pair_1)
print(distance_pair_2)
```
